# Pandas Data Summary

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame for demonstration
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', None],
    'Age': [24, 27, None, 32, 29, 22],
    'Salary': [50000, 54000, 61000, 80000, None, 45000],
    'Department': ['HR', 'Engineering', 'Engineering', 'HR', 'Marketing', 'Marketing']
}
df = pd.DataFrame(data)

# 1. Basic Information (columns, data types, non-null counts)
print("--- Info ---")
df.info()

# 2. Shape of the DataFrame
print(f"\nShape of DataFrame: {df.shape} (Rows, Columns)")

# 3. Descriptive Statistics for numerical columns
print("\n--- Descriptive Statistics (Numerical) ---")
print(df.describe())

# 4. Descriptive Statistics for categorical columns
print("\n--- Descriptive Statistics (Categorical) ---")
print(df.describe(include=[object]))

# 5. Missing values count per column
print("\n--- Missing Values ---")
print(df.isnull().sum())

# 6. Value counts for a specific categorical column
print("\n--- Department Value Counts ---")
print(df['Department'].value_counts())
```
