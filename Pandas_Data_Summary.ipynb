# Pandas Data Summary

```python
import pandas as pd

# Load the CSV file
# Replace 'your_file.csv' with the path to your actual CSV file
df = pd.read_csv('your_file.csv')

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
if 'df' in locals():
    # Calculate total missing values and percentage per column
    missing_count = df.isnull().sum()
    missing_percent = 100 * df.isnull().sum() / len(df)

    # Combine results into a table
    missing_table = pd.concat([missing_count, missing_percent], axis=1)
    missing_table.columns = ['Missing Values', '% of Total Values']

    # Filter and display columns that have missing values
    missing_table = missing_table[missing_table['Missing Values'] > 0].sort_values('% of Total Values', ascending=False)

    print(f"The DataFrame has {df.shape[1]} columns.")
    print(f"There are {missing_table.shape[0]} columns that have missing values.\n")

    if not missing_table.empty:
        display(missing_table)
    else:
        print("No missing values found in any column!")
else:
    print("Please run the previous cell first to load your DataFrame 'df'.")
```
