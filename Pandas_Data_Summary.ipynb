# Pandas Data Summary

```python
import pandas as pd
import seaborn as sns

# Load a sample dataset (for demonstration)
df = sns.load_dataset('iris')

# Display the first 5 rows (default)
display(df.head())
```

You can also specify the number of rows you want to display by passing an integer to `.head(n)`. For example, `df.head(10)` will display the first 10 rows.

### Checking for Missing Values

You can identify missing values in your DataFrame using the `.isnull()` method. To get a quick summary of the number of missing values in each column, chain it with `.sum()`.

```python
# Check the number of missing values in each column
missing_summary = df.isnull().sum()
print("Missing values per column:")
print(missing_summary)
```

If you want to check if there are *any* missing values in the entire DataFrame at all, you can use `.isnull().values.any()`:

```python
# Check if there are any missing values in the entire DataFrame
has_missing = df.isnull().values.any()
print(f"Are there any missing values in the DataFrame? {has_missing}")
```

### Dropping Rows with Missing Values

You can remove rows with missing values using the `.dropna()` method. Let's see how it works.

```python
# Drop all rows that contain at least one missing value
df_cleaned = df.dropna()

# Alternatively, to drop rows in-place (modify the original DataFrame):
# df.dropna(inplace=True)

print("Original shape:", df.shape)
print("Shape after dropping missing rows:", df_cleaned.shape)
```

### Dropping Columns with Missing Values

To drop columns containing missing values instead of rows, specify `axis=1` (or `axis='columns'`) inside the `.dropna()` method.

```python
# Drop all columns that contain at least one missing value
df_cols_cleaned = df.dropna(axis=1)

print("Original shape:", df.shape)
print("Shape after dropping columns with missing values:", df_cols_cleaned.shape)
```

### Dropping Columns Based on a Threshold of Missing Values

You can use the `thresh` parameter to set a threshold of minimum **non-missing** values required to keep a column.

For example, if you want to keep columns that have at least 60% non-missing values, you can calculate that threshold and pass it to `.dropna(axis=1, thresh=threshold)`.

```python
# Calculate the threshold (e.g., keep columns with at least 80% non-missing values)
threshold = int(0.8 * len(df))

# Drop columns that do not meet the threshold
df_thresh_cleaned = df.dropna(axis=1, thresh=threshold)

print(f"Required non-missing values to keep a column: {threshold} out of {len(df)}")
print("Original shape:", df.shape)
print("Shape after threshold drop:", df_thresh_cleaned.shape)
```
