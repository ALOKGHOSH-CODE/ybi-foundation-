# Basics of Linear Regression Model

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Generate synthetic data
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

# 2. Initialize and fit the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# 3. Retrieve model parameters (y = wx + b)
coefficient = model.coef_[0][0]
intercept = model.intercept_[0]
print(f"Learned Coefficient (slope w): {coefficient:.4f}")
print(f"Learned Intercept (bias b): {intercept:.4f}")

# 4. Make predictions for plotting the regression line
X_new = np.array([[0], [2]])
y_predict = model.predict(X_new)

# 5. Plot the results
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='blue', label='Data points')
plt.plot(X_new, y_predict, color='red', linewidth=2, label=f'Regression Line (y = {coefficient:.2f}x + {intercept:.2f})')
plt.xlabel('Independent variable (X)')
plt.ylabel('Dependent variable (y)')
plt.title('Basic Linear Regression Demonstration')
plt.legend()
plt.grid(True)
plt.show()
```
