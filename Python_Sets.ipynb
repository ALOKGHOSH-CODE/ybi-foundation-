# Python Sets

```python
# Python Sets Demonstration

# 1. Creating Sets
print("--- Creating Sets ---")
# Sets are unordered collections of unique elements
my_set = {1, 2, 3, 4, 4, 5}  # Duplicates are automatically removed
print(f"Set with duplicates removed: {my_set}")

# Creating an empty set (must use set(), not {})
empty_set = set()
print(f"Type of empty_set: {type(empty_set)}\n")

# 2. Modifying a Set
print("--- Modifying Sets ---")
my_set.add(6)
print(f"After adding 6: {my_set}")
my_set.remove(3)  # Raises KeyError if 3 is not present
print(f"After removing 3: {my_set}")
my_set.discard(10)  # Does not raise an error even if 10 is not present
print(f"After discarding 10: {my_set}\n")

# 3. Set Operations
print("--- Set Operations ---")
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(f"Set A: {set_a}")
print(f"Set B: {set_b}")

# Union
print(f"Union (A | B): {set_a.union(set_b)}")

# Intersection
print(f"Intersection (A & B): {set_a.intersection(set_b)}")

# Difference (elements in A but not in B)
print(f"Difference (A - B): {set_a.difference(set_b)}")

# Symmetric Difference (elements in either A or B, but not both)
print(f"Symmetric Difference (A ^ B): {set_a.symmetric_difference(set_b)}")
```
