# Python Sets

```python
# Example of Set Union (combines elements from both sets, removing duplicates)
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Using the | operator
union_operator = set_a | set_b

# Using the union() method
union_method = set_a.union(set_b)

print("Set A:", set_a)
print("Set B:", set_b)
print("Union (using |):", union_operator)
print("Union (using method):", union_method)
```

```python
# Example of Set Intersection (keeps only elements present in both sets)
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Using the & operator
intersection_operator = set_a & set_b

# Using the intersection() method
intersection_method = set_a.intersection(set_b)

print("Set A:", set_a)
print("Set B:", set_b)
print("Intersection (using &):", intersection_operator)
print("Intersection (using method):", intersection_method)
```
