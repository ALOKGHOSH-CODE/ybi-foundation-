# ANN Classification

### Basic Structure of an ANN in Keras

To build and train a basic Artificial Neural Network (ANN) classifier in Keras, you typically follow these steps:

1. **Define the Model**: Use `keras.Sequential` to stack layers sequentially.
2. **Add Layers**: Use `layers.Dense` for fully connected layers, specifying the number of neurons and the activation function (e.g., `'relu'` for hidden layers, and `'sigmoid'` or `'softmax'` for the output layer).
3. **Compile the Model**: Configure the learning process by specifying the optimizer (e.g., `'adam'`), the loss function (e.g., `'binary_crossentropy'` for binary classification), and metrics to monitor (e.g., `['accuracy']`).

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

# 1. Define the Sequential model structure
model = Sequential([
    # Input layer and first hidden layer (e.g., 16 neurons, expecting 8 input features)
    Dense(16, activation='relu', input_shape=(8,)),

    # Second hidden layer
    Dense(8, activation='relu'),

    # Output layer (1 neuron with Sigmoid activation for binary classification)
    Dense(1, activation='sigmoid')
])

# 2. Compile the model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Display the model summary to see the architecture and parameters
model.summary()
```

### Training the Model using `.fit()`

To train the model, you use the `fit()` method by passing your training features (`X`) and targets (`y`). Here is an example using randomly generated dummy data to represent 1,000 samples with 8 features each.

```python
import numpy as np

# Generate dummy training data (1000 samples, 8 features each)
X_train = np.random.random((1000, 8))
# Generate dummy binary targets (0 or 1)
y_train = np.random.randint(2, size=(1000, 1))

# Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=10,            # Number of times to iterate over the entire dataset
    batch_size=32,        # Number of samples per gradient update
    validation_split=0.2  # Fraction of the training data to be used as validation data
)
```

### Visualizing Training and Validation Loss

We can extract the loss metrics from the `history.history` dictionary and plot them using `matplotlib` to analyze how the model converges over the epochs.

```python
import matplotlib.pyplot as plt

# Plot training & validation loss values
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss', color='blue')
plt.plot(history.history['val_loss'], label='Validation Loss', color='orange')
plt.title('Model Loss Over Epochs')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```

### Saving the Trained Model

You can save the entire model (including its architecture, weights, and compilation training configuration) using `model.save()`. The standard and recommended format in Keras 3 is the `.keras` extension.

```python
# Save the model to a file named 'my_model.keras'
model.save('my_model.keras')
print("Model saved successfully as 'my_model.keras'")
```
