# Python Dictionary

```python
# Creating a dictionary
my_dict = {
    'name': 'Alice',
    'age': 25,
    'city': 'New York',
    'skills': ['Python', 'Data Science']
}
print("Original Dictionary:", my_dict)

# Accessing values
print("Name:", my_dict['name'])
print("Skills:", my_dict.get('skills'))

# Modifying/Adding elements
my_dict['age'] = 26  # Update existing key
my_dict['profession'] = 'Engineer'  # Add new key-value pair
print("\nUpdated Dictionary:", my_dict)

# Dictionary methods
print("\nKeys:", list(my_dict.keys()))
print("Values:", list(my_dict.values()))
print("Items:", list(my_dict.items()))
```
