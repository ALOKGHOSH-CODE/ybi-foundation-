# Python Loops

```python
# 1. Basic 'for' loop using range()
print("--- Basic for loop (0 to 4) ---")
for i in range(5):
    print(f"Iteration {i}")

# 2. Iterating over a list
print("\n--- Iterating over a list ---")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")

# 3. Basic 'while' loop
print("\n--- Basic while loop ---")
count = 0
while count < 3:
    print(f"Count is {count}")
    count += 1

# 4. Loop control statements (break and continue)
print("\n--- Loop with break and continue ---")
for number in range(1, 6):
    if number == 2:
        continue  # Skip 2
    if number == 5:
        break     # Stop loop at 5
    print(f"Number: {number}")
```
